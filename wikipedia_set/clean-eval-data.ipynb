```python
import json

# Filepath to the JSON file (change it to your file path)
json_file_path = "/Users/maingoclanvy/scale-equity-nlp/wikipedia_set/evaluation_dataset.json"

# Load the JSON file
with open(json_file_path, "r") as f:
    data = json.load(f)

# 1. "Options" field should contain exactly 4 elements
invalid_entries = []
valid_data = []

for i, entry in enumerate(data):
    if "options" in entry and len(entry["options"]) == 4:
        valid_data.append(entry)  # Keep valid entries
    else:
        invalid_entries.append((i, entry))  # Collect invalid entries

if invalid_entries:
    print(f"Found {len(invalid_entries)} invalid entries:")
    for index, entry in invalid_entries:
        print(f"Entry {index}: options = {entry.get('options', None)}")
else:
    print("All entries have exactly 4 elements in the 'options' field.")

with open(json_file_path, "w") as f:
    json.dump(valid_data, f, indent=4)

print(f"Removed {len(invalid_entries)} invalid entries and saved the cleaned dataset.")
```
